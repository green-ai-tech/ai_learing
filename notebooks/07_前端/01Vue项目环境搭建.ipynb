# 1. Vue项目环境搭建

## 1.1. 安装node与npm

### (1) 关于JavaScript语言与Nodejs

- nodejs是与python一样的虚拟机，
    - nodejs主要用来执行Javascript语言。
    - python主要用来执行python语言。

- ECMAScript（简称ES）与Javascript语言：
    - ECMAScript 是一套脚本语言的规范标准。它定义了这门语言应该有什么样的语法、数据类型、关键字、控制流、对象模型等。
        - 版本号：通常以 ES + 数字表示，例如 ES5、ES6（也叫 ES2015）、ES7（ES2016）... 直到现在的 ES14、ES15 等。
    - JavaScript 是ECMAScript标准的一种具体实现。

- TypeScript与Javascript
    - TypeScript 是 JavaScript 的超集，它完全遵循 ECMAScript 标准，并额外添加了类型系统。最终 TypeScript 代码会编译成纯 JavaScript 运行。

- 服务器Javascript与浏览器JavaScript：
    - 服务器JavaScript：扩展了服务器端特有 API 的 JavaScript。（在Nodejs运行的JavaScript）
    - 浏览器JavaScript：扩展了服务器端特有 API 的 JavaScript。（在浏览器运行的JavaScript）

### (2) Nodejs与npm

- Node.js的核心作用是让JavaScript代码能够在服务器端（命令行、后台）运行。
    - 之前：JavaScript 只能运行在浏览器里，用来做网页交互（比如按钮点击、表单验证）。
    - 现在：有了 Node.js，开发者可以用 JavaScript 写服务器、命令行工具、桌面应用、甚至嵌入式设备脚本。

- npm 是 Node Package Manager 的缩写，它的核心作用是管理项目依赖的第三方代码包。
    - npm与python中的匹配一样的作用与地位。

### (7) 配置Tailwind样式表

- 打开`tailwind.config.js`文件，使用下面内容替换：

 ```javascript
 /** @type {import('tailwindcss').Config} */
export default {
  content: [
    "./index.html",
    "./src/**/*.{vue,js,ts,jsx,tsx}",
  ],
  theme: {
    extend: {
      colors: {
        primary: '#155EEF', // Dify 风格的主蓝色
        neutral: {
          100: '#F9FAFB',
          200: '#F3F4F6',
          500: '#6B7280',
          700: '#374151',
          900: '#111827',
        }
      },
      fontFamily: {
        sans: ['Inter', 'system-ui', 'sans-serif'],
      }
    },
  },
  plugins: [],
}
 ```

### (8) 引入样式表

- 在src/assets/main.css中引入如下样式表模块：
    - `@tailwind base;`
    - `@tailwind components;`
    - `@tailwind utilities;`
