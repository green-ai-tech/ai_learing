### (2) 把Views组件设置为URL

```javascript
import { createRouter, createWebHistory } from 'vue-router'
import AgentsAppView from '../views/AgentsApp.vue'

const router = createRouter({
  history: createWebHistory(import.meta.env.BASE_URL),
  routes: [
    {
      path: '/agentsapp.view',
      name: 'home',
      component: AgentsAppView,
    },
  ],
})

export default router
```

### (3) 改进App.vue

- 把views组件引用改变为：
    - `<router-view />`

- 使用router显示路由URL

```html
<!-- AppLayout.vue -->
<template>
  <div class="min-h-screen bg-gray-50 font-sans text-gray-900">
    <AppHeader />
    <!-- <Workspace /> -->
     <router-view />
  </div>
</template>

<script setup>
import AppHeader from './components/Header.vue';
// import Workspace from './views/AgentsApp.vue';
// 使用vue-router组件来设置<router-view />显示的路由。
import { useRouter } from 'vue-router';  
const router = useRouter();
router.push('/agentsapp.view');
</script>
```

## 1.2. 切换路由

### (1) 增加一个view路由组件

```html
<template>
  <main class="bg-gray-50 min-h-screen pb-20">
    工作室
  </main>
</template>

<script setup>
</script>

```

### (2) 配置路由

```javascript
import { createRouter, createWebHistory } from 'vue-router'
import AgentsAppView from '../views/AgentsApp.vue'
import KnowledgeBaseView from '../views/knowledgebase.vue'

const router = createRouter({
  history: createWebHistory(import.meta.env.BASE_URL),
  routes: [
    {
      path: '/agentsapp.view',
      name: 'agentsapp',
      component: AgentsAppView,
    },
    {
      path: '/knowledgebase.view',
      name: 'knowledgebase',
      component: KnowledgeBaseView,
    },
  ],
})

export default router

```

### (3) 配置切换连接

- 在导航组件中配置

```html
<router-link to="/agentsapp.view" class="flex items-center gap-2 px-4 py-2 bg-white rounded-lg shadow-sm font-medium text-sm transition-all" active-class="text-blue-600 font-medium">
    <Box size="16" />
    <span>探索</span>
</router-link>
```

### (4) 设置菜单的属性

- 激活属性：
    - `active-class="text-blue-600 font-medium"`

- 悬停属性：
    - `hover:text-red-600`
