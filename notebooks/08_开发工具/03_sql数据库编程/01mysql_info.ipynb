```python
import mysql
from mysql.connector import Error   # 引入错误对象

# 2. 建立数据库连接
connection = mysql.connector.connect(
    host="localhost",   # IP地址
    port=3306,     # 端口
    user="root",   # 用户 
    password="12345678",   # 密码
    charset="utf8mb4"   # 指定编码
)

try:
    # 3. 数据库处理的缓冲：游标：cursor
    cursor = connection.cursor() 
    # 4. 执行SQL语句
    cursor.execute("show databases;")
    # 5. 处理查询结果（json）
    data = cursor.fetchall()
    print(data)
    # 继续执行其他SQL语句
    for table in data:
        print(table[0])
    
    # 切换数据库
    cursor.execute("use  eduagent;")
    
    # 查看表
    cursor.execute("show tables;")
    tables = cursor.fetchall()
    print(tables)
    
    # 查看字段 
    
except Error as e:
    print("数据库操作错误：", str(e))
finally:
    # 释放游标
    cursor.close()
    # 关闭链接
    connection.close()
```
