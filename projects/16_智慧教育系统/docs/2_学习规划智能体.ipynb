# 1. 业务逻辑说明

## 1.1 这个模块解决什么问题？

学习规划智能体根据：
-   教学大纲
-   学生学习目标
-   学生当前基础
-   学习周期
-   每日可用时间
-   薄弱知识点
-   学习偏好
-   作业/错题分析结果

自动生成一份结构化，可执行，可追踪，可调整的学习计划

它解决的问题是：

过去学生不知道：
- 先学什么？
- 后学什么？
- 每天学多少？
- 如何复习？
- 如何根据薄弱点调整计划
- 如何把长期目标拆为每日任务

学习规划智能体要把“模糊学习目标”转化为“明确学习目标”

## 1.2 用户是谁？

主要用户：

1. 学生
    - 输入学习目标
    - 获得每日学习任务
    - 按计划每日学习和大卡

2. 老师
    - 根据教学大纲给学生 生成学习安排
    - 根据学生薄弱点调整教学节奏

3. 教学平台
    - 为不同的学生生成个性化学习途径
    - 与作用批改, 错题分析, 知识库问答模块联动

## 1.3 输入是什么？

核心输入：
```python
{
  "learning_goal":      "学习目标",
  "current_level":      "当前基础",
  "syllabus":           "课程大纲",
  "available_time_per_day": "每日可学习时间",
  "learning_duration":  "学习周期",
  "weak_points":        ["薄弱知识点"],
  "target_type":        "考试 / 项目 / 面试 / 日常学习",
  "learning_preference": "学习偏好",
  "constraints":         "限制条件"
}
```

拓展输入：
```python
{
  "homework_result":  "作业批改结果",
  "mistake_analysis": "错题分析结果",
  "history_plan":     "历史学习计划",
  "completed_tasks":  "已完成任务",
  "teacher_requirement": "教师要求"
}
```

## 1.4 输出是什么？


输出不是普通聊天文本，而是结构化学习计划。

核心结构化输出：
```markdown
{
  "plan_title":   "学习计划标题",
  "plan_summary": "整体规划说明",
  "stage_plan": [
    {
      "stage_name": "阶段名称",
      "duration":   "阶段周期",
      "goal":       "阶段目标",
      "key_topics":   ["核心知识点"],
      "deliverables": ["阶段产出"]
    }
  ],
  "daily_tasks": [
    {
      "day":        "Day 1",
      "topic":      "当天主题",
      "tasks":      ["具体学习任务"],
      "practice":   ["练习任务"],
      "review":     ["复习任务"],
      "estimated_time": "预计时间",
      "deliverable": "当天产出",
      "check_standard": "完成标准"
    }
  ],
  "review_plan": [
    {
      "time":       "复习时间",
      "content":    "复习内容",
      "method":     "复习方式"
    }
  ],
  "evaluation_criteria":    ["评估标准"],
  "adjustment_suggestions": ["调整建议"]
}
```

## 1.5 成功的标准是什么

学习智能体的成功标准：

1. 可执行：
    - 每天任务清楚
    - 时间安排合理
    - 有明确产出

2. 个性化：
    - 根据学生的基础来调整难度
    - 根据薄弱点安排复习计划
    - 根据学生目标安排学习路径

3. 可追踪：
    - 能记录计划
    - 能记录完成度
    - 能查看历史

4. 可调整：
    - 后续根据错题, 作业, 打开情况调整计划

5. 可工程化:
    - 结构化输出json
    - 可以保存到数据库
    - 可以被前端稳定渲染

6. 可联动
   - 可以接入教学大纲
   - 可以接入错题分析
   - 可以接入知识库
   - 可以接入作业批改结果

# 3. 模式选择

Workflow + State + Skill 模式

## 3.1 选择什么模式

- 主模式：Workflow 图模式
- 辅助模式：Skill 模块化能力
- 核心支撑：AgentState 状态管理

## 3.2 为什么选择这个模式

## 3.3 和其他模式相比较

### 1. 单Agent模式

优点：
- 简单
- 开发快

缺点：
- 输出不稳定
- 逻辑都挤在一个 Prompt 里
- 后期不好扩展动态调整、错题分析联动

结论：
适合 Demo，不适合作为 EduAgent 的正式模块。

### 2. ReAct/Tool Agent模式

优点：
- 适合需要频繁调用工具
- 适合搜索、查询、计算、检索类任务

缺点：
- 学习规划的流程相对固定
- 不需要每一步都让模型自由决策
- 可控性不如 Workflow

结论：
可以局部使用 Tool，但不适合作为主模式。

### 3. 交接模式

优点：
- 适合多个角色协作
- 比如规划专家、测评专家、资料推荐专家互相交接

缺点：
- 当前阶段复杂度过高
- 不利于快速落地
- 调试成本高

结论：
后期可以升级，目前不建议。

### 4. Multi-Agent / Supervisor 模式

优点：
- 扩展性强
- 适合多个专业 Agent 协同

缺点：
- 路由、状态同步、结果合并复杂
- 当前学习规划模块还不需要这么重

结论：
平台级后期可以做，现在不做。

### 5. Workflow + State + Skill 模式

优点：
- 流程透明
- 状态可追踪
- 输出稳定
- 容易测试
- 容易接数据库
- 容易和错题分析、作业批改、大纲生成联动

缺点：
- 前期设计稍微重一点

结论：
最适合当前 LearningPlannerAgent。

# 4. Agent设计

## 4.1 整体结构：

```mermaid
flowchart TD

    A[用户输入 / 教学大纲 / 薄弱点 / 历史数据] --> B[Input Parser 输入解析]

    B --> C[AgentState 状态容器]

    C --> D[GoalAnalysisSkill 目标分析]
    D --> E[LevelAnalysisSkill 学生水平分析]
    E --> F[SyllabusBreakdownSkill 知识点拆解]
    F --> G[StagePlanSkill 阶段计划生成]
    G --> H[DailyTaskSkill 每日任务生成]
    H --> I[ReviewPlanSkill 复习计划生成]
    I --> J[EvaluationSkill 评估标准生成]
    J --> K[AdjustmentSuggestionSkill 调整建议生成]

    K --> L[StructuredOutputSkill 结构化输出]

    L --> M[JSON 学习计划]
    M --> N[保存到数据库]
    M --> O[前端页面展示]
    M --> P[后续动态调整]
```

## 4.2 核心 Agent / Skill / Tool 组件设计

### 1. LearningPlannerAgent 主 Agent

职责：负责整体流程编排，不负责具体细节生成。

主要负责：

1. 接收用户输入
2. 初始化 AgentState
3. 按顺序调用各个 Skill
4. 汇总中间结果
5. 输出最终学习计划

### 2. InputParserSkill 输入解析

职责：把前端表单输入、历史数据、错题分析结果整理成统一格式。

输入：
```json
{
  "learning_goal":          "掌握 RAG 项目开发",
  "current_level":          "有 Python 基础",
  "available_time_per_day": "2小时",
  "learning_duration":      "4周",
  "weak_points":            ["向量数据库", "LangChain"]
}
```

输出：
```json
{
  "normalized_goal":  "4周内掌握 RAG 工程开发并完成一个小项目",
  "normalized_level": "Python基础较好，AI工程经验较弱",
  "time_budget":      {"daily_hours": 2,"total_weeks": 4},
  "planning_constraints": ["偏工程实践","每日任务不宜过重"]
}
```

### 3. GoalAnalysisSkill 目标分析

职责：分析用户学习目标属于哪一类，以及最终应该达到什么标准。

目标类型：

输入示例：

    "能独立实现最小 RAG Demo",
    "能解释召回、重排、引用来源"

}
```

### 4. LevelAnalysisSkill 学生水平分析

职责：根据当前基础、历史学习记录、错题结果判断学生水平。

判断维度：

## 4.3 AgentState 状态设计

AgentState 是 LearningPlannerAgent 的核心。
它负责保存：

1. 用户输入
2. 历史上下文
3. 中间分析结果
4. 各 Skill 的输出
5. 最终学习计划
6. 后续动态调整需要的数据

````python
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field


class TimeBudget(BaseModel):
    daily_hours: Optional[float] = Field(default=None, description="每日可学习小时数")
    total_days: Optional[int]    = Field(default=None, description="总学习天数")
    total_weeks: Optional[int]   = Field(default=None, description="总学习周数")


class LearningPlannerState(BaseModel):
    # 1. 原始输入
    user_id: Optional[int] = None
    learning_goal: str = ""
    current_level: str = ""
    syllabus: Optional[str] = None
    available_time_per_day: str = ""
    learning_duration: str = ""
    target_type: str = ""
    learning_preference: Optional[str] = None
    constraints: Optional[str] = None

    # 2. 可选上下文
    weak_points: List[str] = Field(default_factory=list)
    homework_result: Optional[Dict[str, Any]] = None
    mistake_analysis: Optional[Dict[str, Any]] = None
    history_plan: Optional[Dict[str, Any]] = None
    completed_tasks: List[Dict[str, Any]] = Field(default_factory=list)

    # 3. 输入标准化结果
    normalized_goal: Optional[str] = None
    normalized_level: Optional[str] = None
    time_budget: Optional[TimeBudget] = None
    planning_constraints: List[str] = Field(default_factory=list)

    # 4. Skill 中间结果
    goal_analysis: Optional[Dict[str, Any]] = None
    level_analysis: Optional[Dict[str, Any]] = None
    knowledge_units: List[Dict[str, Any]] = Field(default_factory=list)
    stage_plan: List[Dict[str, Any]] = Field(default_factory=list)
    daily_tasks: List[Dict[str, Any]] = Field(default_factory=list)
    review_plan: List[Dict[str, Any]] = Field(default_factory=list)
    evaluation_criteria: List[Dict[str, Any]] = Field(default_factory=list)
    adjustment_suggestions: List[Dict[str, Any]] = Field(default_factory=list)

    # 5. 最终输出
    plan_title: Optional[str] = None
    plan_summary: Optional[str] = None
    final_plan: Optional[Dict[str, Any]] = None

    # 6. 工程辅助字段
    status: str = "pending"
    error_message: Optional[str] = None
````

## 4.4 学习规划智能体自动化流程设计

```mermaid
flowchart TD

    A[用户进入学习规划智能体] --> B{用户操作类型}

    B -->|新建学习计划| C[填写学习目标与学习背景]
    B -->|基于历史计划调整| D[选择历史学习计划]
    B -->|基于错题/作业生成| E[选择错题分析或作业批改结果]
    B -->|今日任务续学| F[读取当前进行中的学习计划]

    C --> G[InputParserSkill 输入解析]
    D --> H[HistoryContextSkill 历史计划读取]
    E --> I[MistakeContextSkill 错题/作业上下文读取]
    F --> J[ProgressContextSkill 学习进度读取]

    G --> K[初始化 LearningPlannerState]
    H --> K
    I --> K
    J --> K

    K --> L[ContextMergeSkill 上下文合并]

    L --> M{是否缺少关键信息?}

    M -->|缺少| N[ClarificationSkill 生成补充问题]
    N --> O[前端提示用户补充信息]
    O --> G

    M -->|完整| P[GoalAnalysisSkill 学习目标分析]

    P --> Q[LevelAnalysisSkill 学生水平分析]

    Q --> R[WeaknessAnalysisSkill 薄弱点分析]

    R --> S{是否存在教学大纲?}

    S -->|有| T[SyllabusBreakdownSkill 大纲拆解]
    S -->|无| U[KnowledgeMapGenerationSkill 自动生成知识地图]

````python
# Cell 6：调用 DeepSeek API

import requests

DEEPSEEK_API_KEY = "sk-a81b0557212749e48134da69387d62b5"
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1/chat/completions"

MODEL_NAME = "deepseek-chat"
# 也可以用 deepseek-reasoner


def call_deepseek(system_prompt: str, user_prompt: str) -> str:

    headers = {
        "Authorization": f"Bearer {DEEPSEEK_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "model": MODEL_NAME,
        "messages": [
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],
        "temperature": 0.2,
        "max_tokens": 8192
    }

    response = requests.post(
        DEEPSEEK_BASE_URL,
        headers=headers,
        json=payload,
        timeout=180
    )

    response.raise_for_status()

    data = response.json()

    return data["choices"][0]["message"]["content"]


raw_output = call_deepseek(
    SYSTEM_PROMPT,
    user_prompt
)

print(raw_output[:2000])
````

````python
# Cell 7：JSON 清洗与解析

import re
import json


def extract_json(text: str) -> Dict[str, Any]:
    """
    尽量从模型输出中提取 JSON。
    用于处理模型偶尔输出 ```json 的情况。
    """
    text = text.strip()

    text = re.sub(r"^```json", "", text)
    text = re.sub(r"^```", "", text)
    text = re.sub(r"```$", "", text)
    text = text.strip()

    start = text.find("{")
    end = text.rfind("}")

    if start == -1 or end == -1:
        raise ValueError("模型输出中没有找到 JSON 对象")

    json_text = text[start:end + 1]
    return json.loads(json_text)


parsed_output = extract_json(raw_output)
parsed_output.keys()
````
